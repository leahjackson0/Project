```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/dataset.csv")


if len(df.columns) == 1:
    df = pd.read_csv("../data/dataset.csv", sep=";")


df.columns = df.columns.str.strip().str.replace('"', '').str.replace("'", "")

print("Successfully loaded dataset!")
print("Columns found:", list(df.columns[:5]), "... Total columns:", len(df.columns))
df.head()
```

```python

sns.set_theme(style="whitegrid")

# Verify 'Target' exists before plotting
target_col = [col for col in df.columns if col.lower() == 'target']

if target_col:
    actual_target_name = target_col[0]
    
    # Chart 1: Outcome Distribution
    plt.figure(figsize=(8, 5))
    sns.countplot(data=df, x=actual_target_name, palette='Set2')
    plt.title("Distribution of Student Outcomes")
    plt.xlabel("Outcome")
    plt.ylabel("Count")
    plt.show()

    # Chart 2: 1st Sem Grade vs Outcome (if 1st sem grade exists)
    grade_cols = [c for c in df.columns if '1st sem' in c.lower() and 'grade' in c.lower()]
    if grade_cols:
        plt.figure(figsize=(8, 5))
        sns.boxplot(data=df, x=actual_target_name, y=grade_cols[0], palette='Set2')
        plt.title("1st Semester Grades by Student Outcome")
        plt.show()
else:
    print("Could not find 'Target' column. Available columns are:")
    print(df.columns.tolist())
```
